Pandas:
1) Series : one dimensional array(array like object). it is similar to column in table
2) DataFrame : two dimensional array

```python
import pandas as pd
```

```python
import pandas as pd
data=[1,2,3,4,5]
series= pd.Series(data)
print("Series \n", series)
print(type(series))
```

```python
##cerate a series from dictionary
data={'a':1, 'b':2, 'c':3}
dc_series= pd.Series(data)
print(dc_series)
```

```python
data=[10,20,30]
index=['a','b','c']
pd.Series(data,index=index)
```

```python
##dataframe:two dimensional
##create a dataframe from a dictionary of list
data={
    'Name':['Shre','ya','chhaya'],
    'Age':[20,25,30],
    'City':['Ahmedabad','dwarka','goa']
}
df=pd.DataFrame(data)
print(df)
print(type(df))
```

```python
import numpy as np
np.array(df)
```

```python
##create a dataframe from a list of dictionary
data=[
    {'name':'shr','age':30,'city':'banglore'},
    {'name':'ya_','age':30,'city':'banglore'},
]
df=pd.DataFrame(data)
print(df)
```
