```python
##base class
class Animal:
    def speak(self):
        return "Sound of animals"

##derived class1
class Dog(Animal):
    def speak(self):
        return "woo"

#derived class
class Cat(Animal):
    def speak(self):
        return "me"

def ani_speak(animal):
    print(animal.speak())

dog=Dog()
print(dog.speak())
cat=Cat()
print(cat.speak())
ani_speak(dog)
ani_speak(cat)

```

```python
##polymorphism with abstract base class
from abc import ABC,abstractmethod

##define abstract class
class Vehicle(ABC):
    def start_engine(self):
        pass

##drived class
class Car(Vehicle):
    def start_engine(self):
        return "Car engine started"

##derived class 2
class Motor(Vehicle):
    def start_engine(self):
        return "started"

##create object
def start_vehicle(vehicle):
    print(vehicle.start_engine())
car = Car()
motor = Motor()

start_vehicle(car)
```
