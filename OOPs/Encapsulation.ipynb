```python
#getter and setter method
#public, private, protected variables or access modifier

class Person:
    def __init__(self,name,age):
        self.name= name         ##public variable: we can call and define the variable from outside of the class
        self.age=age

def get_name(person):
    return person.name

person=Person("shreya",22)
get_name(person)
```

```python
#getter and setter method
#public, private, protected variables or access modifier

class Person:
    def __init__(self,name,age,gender):
        self.__name= name         ##private variable: we can not call and define the variable from outside of the class
        self.__age=age
        self.gender=gender

def get_name(person):
    return person.__name


person=Person("shreya",22,"female")
get_name(person)

```

```python
#getter and setter method
#public, private, protected variables or access modifier

class Person:
    def __init__(self,name,age,gender):
        self._name= name         ##protected variable: only access by derived class
        self._age=age
        self.gender=gender
class Employee(Person):
    def __init__(self,name,age,gender):
        super().__init__(name,age,gender)

employee=Employee("shre",23,"female")
print(employee._name)
```

```python
##encapsulation with getter and setter
class Person:
    def __init__(self,name,age):
        self.__name= name 
        self.__age=age
    
    #getter method
    def get_name(self):  ##retriving the data
        return self.__name
    #setter method
    def set_name(self,name):  ##changing the name itself
        self.__name=name
    
    #getter method
    def get_age(self):  ##retriving the data
        return self.__age
    #setter method
    def set_age(self,age):  ##changing the name itself
        if age>0:
            self.__age= age
        else:
            print("age cannot be negative")
    
person=Person("shr",22)
##access and modify private variable
print(person.get_name())
person.set_age(35)
print(person.get_age())
person.set_age(-5)
```
