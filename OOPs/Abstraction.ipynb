```python
##showing only necessary features of an obj
from abc import ABC, abstractmethod

##Abstract base class
class Vehicle(ABC):
    def drive(self):
        print("vehicle")
    @abstractmethod
    def start_eng(self):
        pass

class Car(Vehicle):
    def start_eng(self):
        print("car eng started")
    
def operate_vehicle(vehicle):
    vehicle.start_eng()

car=Car()
operate_vehicle(car)
```
