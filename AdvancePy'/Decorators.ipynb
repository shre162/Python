```python
##clouser : in clouse method,anything is defined outside from the method, we can also access that method also.

def main_welcome(msg):
    #msg="welcome"
    def sub_welcome_method():
        print("welcome to the...")
        print(msg)
        print("please")
    return sub_welcome_method() ##in closure, return type is always sub method.
```

```python
main_welcome("shreya")
```

```python
##clouser : in clouse method,anything is defined outside from the method, we can also access that method also.

def main_welcome(func):
    #msg="welcome"
    def sub_welcome_method():
        print("welcome to the...")
        func("hello")
        print("please")
    return sub_welcome_method() ##in closure, return type is always sub method.
```

```python
main_welcome(print)
```

```python
##decorators
##clouser : in clouse method,anything is defined outside from the method, we can also access that method also.

def main_welcome(func):
    #msg="welcome"
    def sub_welcome_method():
        print("welcome to the...")
        func()
        print("please")
    return sub_welcome_method() ##in closure, return type is always sub method.
```

```python
def coure_introduction():
    print("this is advance")
coure_introduction()
```

```python
main_welcome(coure_introduction)
```

```python
@main_welcome
def coure_introduction():
    print("this is advance")
```

```python
##decorators

def my_decorator(func):
    def wrapper():
        print("hii")
        func()
        print("HIII")
    return wrapper
```

```python
@my_decorator
def say_hoo():
    print("Hello")
```

```python
say_hoo()
```
