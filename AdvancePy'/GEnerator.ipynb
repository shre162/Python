```python
def my_generator():
    yield 1
    yield 2
    yield 3
```

```python
def read_large_file(file_path):
    with open(file_path,'r') as file:
        for line in file:
            yield line

file_path='LargeFile.txt'
for line in read_large_file(file_path):
    print(line.strip())
```
